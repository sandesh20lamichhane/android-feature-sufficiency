```python
# Bootstrap -- see environment/colab_bootstrap.md for the full version
import os
os.environ.setdefault('AFS_DATA_ROOT', '/content/drive/MyDrive/afs-data')
os.environ.setdefault('AFS_SCRATCH', '/content/afs-scratch')
from afs.paths import Paths
from afs.config import resolve_experiment
from afs.pipeline import run_experiment
paths = Paths.create()
print('data root:', paths.data_root)
```

```python
res = run_experiment(resolve_experiment('02_reproduce_baseline'), paths)
import pandas as pd
df = pd.DataFrame(res['arms']).T[['n_features','roc_auc','accuracy','f1']]
display(df.round(4))
```

### Gate check


```python
acc = {k: v['accuracy'] for k, v in res['arms'].items()}
gap = max(acc.values()) - min(acc.values())
print(f'accuracy spread: {gap*100:.2f} pp')
print('PREMISE HOLDS -- proceed to 03' if gap < 0.02 else 'PREMISE FAILS -- stop and reconsider')
```
