# 03 · The low-FPR regime

**Week 2 gate.** Same models, no new data. Re-measure where a store-scale
scanner actually operates.

Aggregate accuracy is the wrong instrument: when both arms sit at 97–99%,
"the gap is tiny" is a statement about the metric, not the features.

Differences use a **paired** bootstrap over the same test samples. An unpaired
CI would badly overstate the uncertainty of the difference.


```python
# Bootstrap -- see environment/colab_bootstrap.md for the full version
import os
os.environ.setdefault('AFS_DATA_ROOT', '/content/drive/MyDrive/afs-data')
os.environ.setdefault('AFS_SCRATCH', '/content/afs-scratch')
from afs.paths import Paths
from afs.config import resolve_experiment
from afs.pipeline import run_experiment
paths = Paths.create()
print('data root:', paths.data_root)
```

```python
res = run_experiment(resolve_experiment('03_low_fpr'), paths)
import pandas as pd
cols = ['roc_auc','tpr@fpr=0.1','tpr@fpr=0.01','tpr@fpr=0.001','tpr@fpr=0.0001']
display(pd.DataFrame(res['arms']).T[cols].round(4))
```

### Paired differences vs the permissions-only baseline

If the interval straddles zero, the extra features bought nothing at that operating point — and saying so is the finding.


```python
import pandas as pd
rows = []
for arm, d in res.get('vs_baseline', {}).items():
    for op, s in d.items():
        rows.append({'arm': arm, 'operating_point': op, **s})
display(pd.DataFrame(rows).round(4))
```

### Decision point

Know in advance which paper you are writing:

- **Gap opens at low FPR** → the headline is that aggregate accuracy hid it.
- **Gap stays closed here AND under evasion cost** → permissions really are
  sufficient. Much stronger and more surprising than planned, and needs far
  more than two datasets to defend.
